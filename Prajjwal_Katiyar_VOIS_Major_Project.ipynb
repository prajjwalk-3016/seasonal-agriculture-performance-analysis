```python
df['Rainfall_mm'] = df.groupby('Season')['Rainfall_mm'].transform(
    lambda x: x.fillna(x.median())
)

df['Soil_Moisture_pct'] = df.groupby('Season')['Soil_Moisture_pct'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df = df.dropna(subset=['Yield_Tonnes_Ha'])
```

**Season-wise Yield**

```python
import matplotlib.pyplot as plt

season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_yield.index, season_yield.values)

plt.title('Average Yield Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Season-wise Profit**

```python
import matplotlib.pyplot as plt

season_profit = df.groupby('Season')['Profit_INR'].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_profit.index, season_profit.values)

plt.title('Average Profit Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.axhline(0, linewidth=0.8)
plt.tight_layout()
plt.show()
```

**Season-wise Revenue**

```python
import matplotlib.pyplot as plt

season_revenue = df.groupby('Season')['Revenue_INR'].mean()

plt.figure(figsize=(8, 5))
plt.bar(season_revenue.index, season_revenue.values)

plt.title('Average Revenue Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Revenue (INR)')
plt.tight_layout()
plt.show()
```

**Season-wise Environmental Conditions**

```python
import matplotlib.pyplot as plt

environment = df.groupby('Season')[[
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_Moisture_pct'
]].mean()

environment.plot(kind='bar', figsize=(12, 6))

plt.title('Average Environmental Conditions Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.legend(title='Environmental Factors')
plt.tight_layout()
plt.show()
```

**Water Usage vs Water Efficiency**

```python
import matplotlib.pyplot as plt

water = df.groupby('Season')[[
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3'
]].mean()

fig, ax1 = plt.subplots(figsize=(8, 5))

ax1.bar(water.index, water['Water_Used_m3'])
ax1.set_xlabel('Season')
ax1.set_ylabel('Average Water Used (m³)')

ax2 = ax1.twinx()
ax2.plot(
    water.index,
    water['Water_Efficiency_t_per_1000m3'],
    marker='o'
)
ax2.set_ylabel('Water Efficiency (Tonnes/1000m³)')

plt.title('Water Usage and Efficiency Across Seasons')
plt.tight_layout()
plt.show()
```

**Crop-wise Yield Across Seasons**

```python
import matplotlib.pyplot as plt

crop_yield = df.groupby(['Season', 'Crop'])['Yield_Tonnes_Ha'].mean().unstack()

crop_yield.plot(kind='bar', figsize=(12, 6))

plt.title('Average Crop Yield Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=0)
plt.legend(title='Crop')
plt.tight_layout()
plt.show()
```

**State-wise Yield + State-wise Profit**

```python
import matplotlib.pyplot as plt

state_data = df.groupby('State')[[
    'Yield_Tonnes_Ha',
    'Profit_INR'
]].mean()

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Yield
axes[0].bar(state_data.index, state_data['Yield_Tonnes_Ha'])
axes[0].set_title('Average Yield by State')
axes[0].set_xlabel('State')
axes[0].set_ylabel('Yield (Tonnes/Ha)')
axes[0].tick_params(axis='x', rotation=45)

# Profit
axes[1].bar(state_data.index, state_data['Profit_INR'])
axes[1].set_title('Average Profit by State')
axes[1].set_xlabel('State')
axes[1].set_ylabel('Profit (INR)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
